```python
"""
=============================================================
  Medical Chatbot  —  Streamlit App
  NLP: TF-IDF Q&A  +  Random Forest Symptom Classifier
  Diseases: 41 original + PCOD + Pneumonia  (43 total)

  Run:
      pip install streamlit pandas scikit-learn numpy
      streamlit run app.py
=============================================================
"""

import re, ast, warnings
import numpy as np
import pandas as pd
import streamlit as st
from sklearn.ensemble import RandomForestClassifier
from sklearn.preprocessing import LabelEncoder
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.metrics.pairwise import cosine_similarity

warnings.filterwarnings("ignore")

# ──────────────────────────────────────────────────────────────────────────────
# 0.  PAGE CONFIG  (must be first Streamlit call)
# ──────────────────────────────────────────────────────────────────────────────
st.set_page_config(
    page_title="MediBot — Medical Assistant",
    page_icon="🏥",
    layout="wide",
    initial_sidebar_state="expanded",
)

# ──────────────────────────────────────────────────────────────────────────────
# 1.  CUSTOM CSS
# ──────────────────────────────────────────────────────────────────────────────
st.markdown("""
<style>
@import url('https://fonts.googleapis.com/css2?family=DM+Serif+Display&family=DM+Sans:wght@300;400;500;600&display=swap');

/* ── Base ─────────────────────────────────────────── */
html, body, [class*="css"] {
    font-family: 'DM Sans', sans-serif;
    background: #f0f4f8;
}
.stApp { background: #f0f4f8; }

/* ── Sidebar ──────────────────────────────────────── */
[data-testid="stSidebar"] {
    background: linear-gradient(160deg, #0f2027, #203a43, #2c5364);
    border-right: none;
}
[data-testid="stSidebar"] * { color: #e8f4fd !important; }
[data-testid="stSidebar"] .stMarkdown h1,
[data-testid="stSidebar"] .stMarkdown h2,
[data-testid="stSidebar"] .stMarkdown h3 {
    font-family: 'DM Serif Display', serif;
    color: #7ecfff !important;
}
[data-testid="stSidebar"] hr { border-color: rgba(126,207,255,0.3); }

/* ── Chat bubbles ─────────────────────────────────── */
.chat-container {
    display: flex;
    flex-direction: column;
    gap: 16px;
    padding: 12px 0;
}
.msg-row { display: flex; align-items: flex-start; gap: 10px; }
.msg-row.user  { flex-direction: row-reverse; }

.avatar {
    width: 38px; height: 38px;
    border-radius: 50%;
    display: flex; align-items: center; justify-content: center;
    font-size: 18px;
    flex-shrink: 0;
}
.avatar.bot  { background: linear-gradient(135deg,#2c5364,#203a43); }
.avatar.user { background: linear-gradient(135deg,#6c63ff,#48c6ef); }

.bubble {
    max-width: 72%;
    padding: 14px 18px;
    border-radius: 18px;
    line-height: 1.6;
    font-size: 0.93rem;
    box-shadow: 0 2px 8px rgba(0,0,0,.08);
}
.bubble.bot {
    background: #ffffff;
    border-top-left-radius: 4px;
    color: #1a2a3a;
    border-left: 3px solid #2c5364;
}
.bubble.user {
    background: linear-gradient(135deg,#6c63ff,#48c6ef);
    border-top-right-radius: 4px;
    color: white;
    text-align: right;
}

/* ── Cards ────────────────────────────────────────── */
.info-card {
    background: white;
    border-radius: 14px;
    padding: 18px 22px;
    margin: 8px 0;
    box-shadow: 0 2px 10px rgba(0,0,0,.07);
    border-left: 4px solid #2c5364;
}
.info-card h4 { margin: 0 0 6px; color: #2c5364; font-size: 0.85rem; letter-spacing:.05em; text-transform:uppercase; }
.info-card p  { margin: 0; color: #334; font-size: 0.92rem; }

/* ── Severity badge ───────────────────────────────── */
.sev-badge {
    display: inline-block;
    padding: 4px 14px;
    border-radius: 20px;
    font-weight: 600;
    font-size: 0.82rem;
    margin: 4px 0;
}
.sev-high   { background:#ffe5e5; color:#c0392b; }
.sev-mod    { background:#fff8e1; color:#b8860b; }
.sev-low    { background:#e8f8e8; color:#27ae60; }

/* ── Header ───────────────────────────────────────── */
.hero-header {
    background: linear-gradient(135deg,#0f2027 0%,#203a43 50%,#2c5364 100%);
    border-radius: 18px;
    padding: 28px 36px;
    margin-bottom: 22px;
    display: flex; align-items: center; gap: 18px;
}
.hero-header h1 {
    font-family: 'DM Serif Display', serif;
    color: #fff;
    font-size: 2rem;
    margin: 0;
}
.hero-header p  { color: #a8d8f0; margin: 4px 0 0; font-size: .95rem; }

/* ── Input area ───────────────────────────────────── */
.stTextInput input {
    border: 2px solid #d0dce8 !important;
    border-radius: 12px !important;
    padding: 12px 16px !important;
    font-size: 0.95rem !important;
    background: #fff !important;
    transition: border .2s;
}
.stTextInput input:focus {
    border-color: #2c5364 !important;
    box-shadow: 0 0 0 3px rgba(44,83,100,.15) !important;
}
button[kind="primary"] {
    background: linear-gradient(135deg,#2c5364,#203a43) !important;
    border: none !important;
    border-radius: 12px !important;
    font-weight: 600 !important;
}

/* ── Quick-pills ──────────────────────────────────── */
.pill-row { display: flex; flex-wrap: wrap; gap: 8px; margin: 10px 0; }
.pill {
    background: #e8f0fe;
    color: #2c5364;
    border: 1px solid #c5d8e8;
    border-radius: 20px;
    padding: 5px 14px;
    font-size: 0.82rem;
    cursor: pointer;
    font-weight: 500;
}

/* ── Disclaimer ───────────────────────────────────── */
.disclaimer {
    background:#fff8e1;
    border:1px solid #f9c74f;
    border-radius:10px;
    padding:10px 16px;
    font-size:0.8rem;
    color:#7d6608;
    margin-top:8px;
}
</style>
""", unsafe_allow_html=True)


# ──────────────────────────────────────────────────────────────────────────────
# 2.  EXTRA DISEASE DEFINITIONS
# ──────────────────────────────────────────────────────────────────────────────
PCOD_SYMPTOMS = [
    "irregular_periods","acne","weight_gain","excess_hair_growth",
    "fatigue","headache","mood_swings","hair_loss","dark_skin_patches"
]
PNEUMONIA_SYMPTOMS = [
    "cough","high_fever","chest_pain","breathlessness",
    "fatigue","headache","chills","malaise","sweating"
]

EXTRA_DISEASES = {
    "PCOD": {
        "description": "PCOD (Polycystic Ovarian Disease) is a hormonal disorder causing irregular periods, acne, weight gain, and fertility issues in women.",
        "causes": "Insulin resistance, high androgen levels, genetic factors, hormonal imbalance.",
        "risk_factors": "Obesity, family history, sedentary lifestyle, diabetes.",
        "diagnosis": "Hormonal blood tests, transvaginal ultrasound, medical history review.",
        "medications": "['Oral contraceptives', 'Metformin', 'Anti-androgens (as prescribed)']",
        "home_remedies": "Healthy diet, yoga, stress management, regular sleep.",
        "diet": "Low refined sugar, high-fiber foods, plenty of fruits and vegetables.",
        "when_to_see_doctor": "If periods are absent for 3+ months, severe pain, rapid weight gain, or infertility concerns.",
        "precautions": ["Consult gynecologist","Track menstrual cycle","Avoid junk food","Regular exercise"],
        "workout": ["30 min brisk walking daily","Yoga and breathing exercises","Strength training 3x per week","Avoid high-impact exercise during flare"],
        "symptoms": PCOD_SYMPTOMS,
    },
    "Pneumonia": {
        "description": "Pneumonia is a lung infection causing inflammation of air sacs, leading to cough, fever, and breathing difficulty.",
        "causes": "Bacterial (Streptococcus pneumoniae), viral infection, aspiration, or weakened immunity.",
        "risk_factors": "Smoking, age (elderly/infants), chronic lung disease, diabetes, immunosuppression.",
        "diagnosis": "Chest X-ray, blood tests, sputum culture, physical examination.",
        "medications": "['Antibiotics (if bacterial)', 'Antipyretics', 'Bronchodilators (as needed)']",
        "home_remedies": "Rest, plenty of fluids, warm liquids, steam inhalation (if allowed by doctor).",
        "diet": "Light nutritious meals, plenty of fluids, avoid heavy oily food.",
        "when_to_see_doctor": "If breathing difficulty worsens, high fever persists, chest pain increases, or lips/fingertips turn blue.",
        "precautions": ["Avoid smoking","Get vaccinated","Avoid crowded places during flu season","Cover mouth while coughing"],
        "workout": ["Complete rest during acute phase","Light breathing exercises post recovery","Gradual return to walking","Avoid strenuous activity until cleared by doctor"],
        "symptoms": PNEUMONIA_SYMPTOMS,
    },
}


# ──────────────────────────────────────────────────────────────────────────────
# 3.  DATA LOADING & INJECTION  (cached)
# ──────────────────────────────────────────────────────────────────────────────
@st.cache_data(show_spinner=False)
def load_and_prepare(base_path: str):
    p = base_path.rstrip("/\\") + "\\"
    def read(f):
        try:    return pd.read_csv(p + f)
        except: return pd.DataFrame()

    dfs = {
        "symptoms":     read("symtoms_df.csv"),
        "description":  read("description.csv"),
        "precautions":  read("precautions_df.csv"),
        "medications":  read("medications.csv"),
        "workout":      read("workout_df.csv"),
        "severity":     read("Symptom-severity.csv"),
        "training":     read("Training.csv"),
        "causes":       read("causes.csv"),
        "risk_factors": read("risk_factors.csv"),
        "diagnosis":    read("diagnosis.csv"),
        "home_remedies":read("home_remedies.csv"),
        "diets":        read("diets.csv"),
        "doctor":       read("when_to_see_doctor.csv"),
    }

    # ── clean ──────────────────────────────────────────────────────────────────
    for key, col in [("symptoms","Disease"),("description","Disease"),
                     ("precautions","Disease"),("medications","Disease"),
                     ("causes","Disease"),("risk_factors","Disease"),
                     ("diagnosis","Disease"),("home_remedies","Disease"),
                     ("diets","Disease"),("doctor","Disease")]:
        if not dfs[key].empty and col in dfs[key].columns:
            dfs[key][col] = dfs[key][col].astype(str).str.strip()
    for key in dfs:
        unnamed = [c for c in dfs[key].columns if "Unnamed" in str(c)]
        dfs[key] = dfs[key].drop(columns=unnamed, errors="ignore")
    if not dfs["workout"].empty and "disease" in dfs["workout"].columns:
        dfs["workout"] = dfs["workout"].rename(columns={"disease":"Disease"})
    if not dfs["workout"].empty and "Disease" in dfs["workout"].columns:
        dfs["workout"]["Disease"] = dfs["workout"]["Disease"].astype(str).str.strip()

    # ── inject PCOD & Pneumonia ────────────────────────────────────────────────
    for disease, info in EXTRA_DISEASES.items():
        syms = info["symptoms"]

        # symptoms rows (chunks of 4)
        for i in range(0, len(syms), 4):
            chunk = syms[i:i+4]
            row = {"Disease": disease}
            for j, s in enumerate(chunk, 1): row[f"Symptom_{j}"] = s
            dfs["symptoms"] = pd.concat([dfs["symptoms"], pd.DataFrame([row])], ignore_index=True)

        dfs["description"] = pd.concat([dfs["description"],
            pd.DataFrame([{"Disease":disease,"Description":info["description"]}])], ignore_index=True)

        prec_row = {"Disease": disease}
        for i,p in enumerate(info["precautions"],1): prec_row[f"Precaution_{i}"] = p
        dfs["precautions"] = pd.concat([dfs["precautions"], pd.DataFrame([prec_row])], ignore_index=True)

        dfs["medications"] = pd.concat([dfs["medications"],
            pd.DataFrame([{"Disease":disease,"Medication":info["medications"]}])], ignore_index=True)

        for w in info["workout"]:
            dfs["workout"] = pd.concat([dfs["workout"],
                pd.DataFrame([{"Disease":disease,"workout":w}])], ignore_index=True)

        for key, col, field in [
            ("causes","Causes","causes"),("risk_factors","Risk_Factors","risk_factors"),
            ("diagnosis","Diagnosis","diagnosis"),("home_remedies","Home_Remedies","home_remedies"),
            ("diets","Diet","diet"),("doctor","When_to_See_Doctor","when_to_see_doctor")
        ]:
            if not dfs[key].empty:
                dfs[key] = pd.concat([dfs[key],
                    pd.DataFrame([{"Disease":disease, col:info[field]}])], ignore_index=True)

        if not dfs["training"].empty:
            tr_row = {c:0 for c in dfs["training"].columns}
            tr_row["prognosis"] = disease
            for s in syms:
                s_clean = s.strip().lower().replace(" ","_")
                if s_clean in tr_row: tr_row[s_clean] = 1
            dfs["training"] = pd.concat([dfs["training"], pd.DataFrame([tr_row])], ignore_index=True)

    return dfs


# ──────────────────────────────────────────────────────────────────────────────
# 4.  ML CLASSIFIER
# ──────────────────────────────────────────────────────────────────────────────
@st.cache_resource(show_spinner=False)
def build_classifier(training_df: pd.DataFrame):
    le = LabelEncoder()
    df = training_df.copy()
    X  = df.drop(columns=["prognosis"], errors="ignore")
    y  = df["prognosis"].astype(str).str.strip()
    feat_cols = X.columns.tolist()
    le.fit(y)
    y_enc = le.transform(y)
    clf = RandomForestClassifier(n_estimators=120, random_state=42)
    clf.fit(X.fillna(0), y_enc)
    return clf, le, feat_cols

def predict_disease(symptoms, clf, le, feat_cols):
    vec = np.zeros(len(feat_cols))
    for s in symptoms:
        sc = s.strip().lower().replace(" ","_")
        if sc in feat_cols: vec[feat_cols.index(sc)] = 1
    proba = clf.predict_proba([vec])[0]
    idx   = np.argmax(proba)
    return le.inverse_transform([idx])[0], round(proba[idx]*100,1)


# ──────────────────────────────────────────────────────────────────────────────
# 5.  Q&A ENGINE
# ──────────────────────────────────────────────────────────────────────────────
@st.cache_resource(show_spinner=False)
def build_qa_engine(_dfs):
    qa_pairs = []

    def add(questions, answer):
        for q in questions: qa_pairs.append((q.lower(), answer))

    def iter_rows(df, dcol, vcol):
        if df.empty or vcol not in df.columns: return
        for _, row in df.iterrows():
            d, v = str(row[dcol]).strip(), str(row[vcol]).strip()
            if d and v and v.lower() != "nan": yield d, v

    for d,v in iter_rows(_dfs["description"],"Disease","Description"):
        add([f"what is {d}",f"tell me about {d}",f"explain {d}",f"describe {d}",f"define {d}"],
            f"📋 **{d}**\n{v}")

    for d,v in iter_rows(_dfs.get("causes",pd.DataFrame()),"Disease","Causes"):
        add([f"what causes {d}",f"cause of {d}",f"reason for {d}",f"why does {d} happen"],
            f"🔬 **Causes of {d}:**\n{v}")

    for d,v in iter_rows(_dfs.get("risk_factors",pd.DataFrame()),"Disease","Risk_Factors"):
        add([f"risk factors for {d}",f"who is at risk of {d}",f"risk of {d}"],
            f"⚠️ **Risk factors for {d}:**\n{v}")

    for d,v in iter_rows(_dfs.get("diagnosis",pd.DataFrame()),"Disease","Diagnosis"):
        add([f"how to diagnose {d}",f"diagnosis of {d}",f"tests for {d}",f"how is {d} detected"],
            f"🩺 **Diagnosis of {d}:**\n{v}")

    for d,v in iter_rows(_dfs["medications"],"Disease","Medication"):
        try: v = ", ".join(ast.literal_eval(v))
        except: pass
        add([f"medication for {d}",f"medicine for {d}",f"treatment for {d}",f"drugs for {d}"],
            f"💊 **Medications for {d}:**\n{v}")

    prec_cols = [c for c in _dfs["precautions"].columns if c.startswith("Precaution_")]
    for _,row in _dfs["precautions"].iterrows():
        d = str(row["Disease"]).strip()
        ps = [str(row[c]) for c in prec_cols if str(row.get(c,"nan")).lower() != "nan"]
        if d and ps:
            add([f"precautions for {d}",f"prevention of {d}",f"how to prevent {d}"],
                f"🛡️ **Precautions for {d}:**\n" + "\n".join(f"• {p}" for p in ps))

    for d,v in iter_rows(_dfs.get("home_remedies",pd.DataFrame()),"Disease","Home_Remedies"):
        add([f"home remedy for {d}",f"natural treatment for {d}",f"remedies for {d}"],
            f"🌿 **Home remedies for {d}:**\n{v}")

    for d,v in iter_rows(_dfs.get("diets",pd.DataFrame()),"Disease","Diet"):
        add([f"diet for {d}",f"what to eat in {d}",f"food for {d}",f"nutrition for {d}"],
            f"🥗 **Diet for {d}:**\n{v}")

    workout_map = {}
    for _,row in _dfs["workout"].iterrows():
        d = str(row["Disease"]).strip()
        w = str(row.get("workout","")).strip()
        if d and w and w.lower() != "nan": workout_map.setdefault(d,[]).append(w)
    for d,wl in workout_map.items():
        add([f"workout for {d}",f"exercise for {d}",f"fitness for {d}"],
            f"🏃 **Workout tips for {d}:**\n" + "\n".join(f"• {w}" for w in wl[:5]))

    for d,v in iter_rows(_dfs.get("doctor",pd.DataFrame()),"Disease","When_to_See_Doctor"):
        add([f"when to see doctor for {d}",f"doctor for {d}",f"should i see a doctor for {d}"],
            f"🏥 **When to see a doctor for {d}:**\n{v}")

    sym_cols = [c for c in _dfs["symptoms"].columns if c.startswith("Symptom_")]
    for _,row in _dfs["symptoms"].iterrows():
        d = str(row["Disease"]).strip()
        ss = [str(row[c]) for c in sym_cols if str(row.get(c,"nan")).lower() not in ("nan","")]
        if d and ss:
            add([f"symptoms of {d}",f"signs of {d}",f"what are symptoms of {d}"],
                f"🤒 **Symptoms of {d}:**\n" + ", ".join(ss))

    vectorizer   = TfidfVectorizer(ngram_range=(1,2), stop_words="english")
    tfidf_matrix = vectorizer.fit_transform([q for q,_ in qa_pairs])
    return qa_pairs, vectorizer, tfidf_matrix

def qa_answer(query, qa_pairs, vectorizer, tfidf_matrix, threshold=0.12):
    q_vec = vectorizer.transform([query.lower()])
    sims  = cosine_similarity(q_vec, tfidf_matrix).flatten()
    idx   = np.argmax(sims)
    if sims[idx] < threshold: return None
    return qa_pairs[idx][1]


# ──────────────────────────────────────────────────────────────────────────────
# 6.  SEVERITY SCORING
# ──────────────────────────────────────────────────────────────────────────────
def score_severity(symptoms, severity_df):
    sev_map = {}
    if not severity_df.empty and "Symptom" in severity_df.columns:
        sev_map = dict(zip(
            severity_df["Symptom"].str.strip().str.lower().str.replace(" ","_"),
            severity_df["weight"].astype(float)
        ))
    scores = {s: sev_map.get(s.strip().lower().replace(" ","_"), 1) for s in symptoms}
    total  = sum(scores.values())
    if total >= 13: level, css = "High", "sev-high"
    elif total >= 6: level, css = "Moderate", "sev-mod"
    else: level, css = "Low", "sev-low"
    return total, level, css


# ──────────────────────────────────────────────────────────────────────────────
# 7.  RESPONSE BUILDER
# ──────────────────────────────────────────────────────────────────────────────
SYMPTOM_RE = re.compile(r"\b(symptom|suffering|feeling|feel|pain|ache|having|experience|i have|i am|i feel)\b", re.I)
GREET_RE   = re.compile(r"\b(hi|hello|hey|howdy|greetings|good\s*(morning|afternoon|evening))\b", re.I)

def build_response(user_input, dfs, clf, le, feat_cols, qa_pairs, vectorizer, tfidf_matrix, context):
    txt = user_input.strip()
    lo  = txt.lower()

    # Greeting
    if GREET_RE.search(lo):
        return (
            "👋 **Hello! I'm MediBot, your Medical Assistant.**\n\n"
            "Here's what I can help with:\n"
            "- 🤒 **Symptom check** — describe symptoms to get a disease prediction\n"
            "- ❓ **Q&A** — ask about causes, diet, medications, remedies, precautions\n"
            "- 🔬 **Specific disease** — 'What is PCOD?' or 'Diet for pneumonia'\n\n"
            "_Try: 'I have fever, headache, and chills'_"
        ), context

    # Farewell
    if re.search(r"\b(bye|goodbye|exit|quit|thank)\b", lo):
        return "👋 Take care and stay healthy! Remember to consult a qualified doctor for medical advice. 🏥", context

    # Symptom-led input
    if SYMPTOM_RE.search(lo) or lo.count(",") >= 1:
        cleaned = re.sub(SYMPTOM_RE, "", txt)
        syms = [s.strip() for s in re.split(r"[,;]|\band\b", cleaned) if s.strip() and len(s.strip()) > 2]
        if not syms:
            return "Please list symptoms separated by commas.\n_Example: fever, headache, chills_", context

        disease, conf = predict_disease(syms, clf, le, feat_cols)
        total, level, css = score_severity(syms, dfs["severity"])
        context["disease"] = disease
        context["symptoms"] = syms

        parts = [
            f"🩺 **Predicted Condition: {disease}**  _(confidence: {conf}%)_\n",
            f"**Severity:** {level} (score: {total})\n",
            "\n_You can now ask me about this condition — e.g., 'What are the precautions?' or 'Diet recommendations?'_"
        ]
        return "\n".join(parts), context

    # Q&A — resolve pronoun "its" / "this disease" to last predicted disease
    query = txt
    if context.get("disease"):
        query = re.sub(r"\bits\b|\bthis disease\b|\bthis condition\b",
                       context["disease"], query, flags=re.I)

    ans = qa_answer(query, qa_pairs, vectorizer, tfidf_matrix)
    if ans:
        return ans, context

    # Try appending last disease to query
    if context.get("disease"):
        ans2 = qa_answer(f"{query} {context['disease']}", qa_pairs, vectorizer, tfidf_matrix)
        if ans2: return ans2, context

    return (
        "🤔 I'm not sure about that. Try asking:\n"
        "- _'What causes diabetes?'_\n"
        "- _'Diet for PCOD'_\n"
        "- _'Precautions for malaria'_\n"
        "- Or describe your symptoms: _'I have fever and chills'_"
    ), context


# ──────────────────────────────────────────────────────────────────────────────
# 8.  SIDEBAR
# ──────────────────────────────────────────────────────────────────────────────
with st.sidebar:
    st.markdown("## 🏥 MediBot")
    st.markdown("*AI-powered Medical Assistant*")
    st.markdown("---")

    base_path = st.text_input(
        "📁 Dataset folder path",
        value=r"C:\Users\Varalakshmi\Desktop\new_new_project\chatbot_dataset",
        help="Path to folder containing all CSV files"
    )

    st.markdown("---")
    st.markdown("### 📚 Supported Topics")
    topics = ["🤒 Symptoms → Disease", "🔬 Causes", "⚠️ Risk Factors",
              "🩺 Diagnosis", "💊 Medications", "🛡️ Precautions",
              "🌿 Home Remedies", "🥗 Diet", "🏃 Workout", "🏥 When to See Doctor"]
    for t in topics: st.markdown(f"- {t}")

    st.markdown("---")
    st.markdown("### 🦠 43 Diseases Covered")
    st.markdown("Including **PCOD** and **Pneumonia** ✅")

    st.markdown("---")
    st.markdown("### 💡 Sample Queries")
    samples = [
        "I have fever, headache, chills",
        "What causes PCOD?",
        "Diet for diabetes",
        "Precautions for malaria",
        "When to see doctor for pneumonia",
    ]
    for s in samples:
        st.markdown(f"• _{s}_")

    st.markdown("---")
    if st.button("🗑️ Clear Chat"):
        st.session_state.messages  = []
        st.session_state.context   = {}
        st.rerun()

    st.markdown("---")
    st.markdown(
        "<small>⚠️ For informational purposes only.<br>Always consult a qualified doctor.</small>",
        unsafe_allow_html=True
    )


# ──────────────────────────────────────────────────────────────────────────────
# 9.  MAIN  —  load models
# ──────────────────────────────────────────────────────────────────────────────
with st.spinner("🔄 Loading datasets and training models…"):
    try:
        dfs = load_and_prepare(base_path)
        has_data = not dfs["training"].empty and not dfs["description"].empty
    except Exception as e:
        st.error(f"Error loading data: {e}")
        has_data = False

if not has_data:
    st.error("⚠️ Could not load datasets. Please check the folder path in the sidebar.")
    st.stop()

with st.spinner("🤖 Building AI models…"):
    clf, le, feat_cols = build_classifier(dfs["training"])
    qa_pairs, vectorizer, tfidf_matrix = build_qa_engine(dfs)


# ──────────────────────────────────────────────────────────────────────────────
# 10.  CHAT UI
# ──────────────────────────────────────────────────────────────────────────────

# ── Hero header ───────────────────────────────────────────────────────────────
st.markdown("""
<div class="hero-header">
  <div style="font-size:3rem">🏥</div>
  <div>
    <h1>MediBot Medical Assistant</h1>
    <p>Symptom checker · Disease Q&A · 43 diseases · NLP-powered</p>
  </div>
</div>
""", unsafe_allow_html=True)

# ── Session state ──────────────────────────────────────────────────────────────
if "messages" not in st.session_state:
    st.session_state.messages = []
    st.session_state.context  = {}
    # welcome message
    st.session_state.messages.append({
        "role": "bot",
        "text": (
            "👋 **Hello! I'm MediBot.**\n\n"
            "I can help you with:\n"
            "- 🤒 **Symptom check** — describe your symptoms for a diagnosis prediction\n"
            "- ❓ **Medical Q&A** — ask about any of the 43 diseases\n\n"
            "_Try: 'I have fever, headache, and chills'  or  'What causes PCOD?'_"
        )
    })

# ── Render chat history ────────────────────────────────────────────────────────
chat_html = '<div class="chat-container">'
for msg in st.session_state.messages:
    role  = msg["role"]
    text  = msg["text"]
    av    = "🤖" if role == "bot" else "👤"
    # convert markdown bold to html for bubble
    text_html = text.replace("**","<b>",1)
    # simple markdown → html
    import html as htmlmod
    text_safe = htmlmod.escape(text)
    # restore bold
    import re as _re
    text_safe = _re.sub(r'\*\*(.+?)\*\*', r'<b>\1</b>', text_safe)
    text_safe = text_safe.replace("\n","<br>")
    text_safe = _re.sub(r'_(.+?)_', r'<i>\1</i>', text_safe)
    chat_html += f"""
    <div class="msg-row {role}">
      <div class="avatar {role}">{av}</div>
      <div class="bubble {role}">{text_safe}</div>
    </div>"""
chat_html += "</div>"
st.markdown(chat_html, unsafe_allow_html=True)

# ── Quick-question pills ───────────────────────────────────────────────────────
st.markdown("---")
st.markdown("**Quick questions:**")
quick = [
    "What is PCOD?", "Diet for diabetes", "Causes of dengue",
    "Precautions for tuberculosis", "Symptoms of pneumonia",
    "When to see doctor for malaria"
]
cols = st.columns(len(quick))
for i, q in enumerate(quick):
    if cols[i].button(q, key=f"pill_{i}", use_container_width=True):
        st.session_state._quick_input = q

# ── Input box ──────────────────────────────────────────────────────────────────
col1, col2 = st.columns([5, 1])
with col1:
    user_input = st.text_input(
        "Your message",
        label_visibility="collapsed",
        placeholder="Type symptoms or ask a question… e.g. 'I have fever, cough, chills'",
        key="user_input"
    )
with col2:
    send = st.button("Send ➤", type="primary", use_container_width=True)

# handle quick pill click
if hasattr(st.session_state, "_quick_input"):
    user_input = st.session_state._quick_input
    del st.session_state._quick_input
    send = True

# ── Process message ────────────────────────────────────────────────────────────
if send and user_input.strip():
    st.session_state.messages.append({"role":"user","text":user_input})
    response, st.session_state.context = build_response(
        user_input, dfs, clf, le, feat_cols,
        qa_pairs, vectorizer, tfidf_matrix,
        st.session_state.context
    )
    st.session_state.messages.append({"role":"bot","text":response})
    st.rerun()

# ── Disclaimer ─────────────────────────────────────────────────────────────────
st.markdown("""
<div class="disclaimer">
  ⚠️ <b>Medical Disclaimer:</b> MediBot is for informational purposes only and does not constitute medical advice.
  Always consult a qualified healthcare professional for diagnosis and treatment.
</div>
""", unsafe_allow_html=True)
```
